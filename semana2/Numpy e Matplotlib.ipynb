```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# A partir de uma lista
meu_array = np.array([1, 2, 3, 4, 5])
print("Meu array:", meu_array)

# Criando um arrau de zeros
zeros = np.zeros((2, 3)) # Matriz 2x3 de zeros
print("Array de zeros:\n", zeros)

# Criando um array com um intervalo
intervalo = np.arange(0, 11, 2)
print(f"Array com intervalo:\n {intervalo}")
```

```python
array_a = np.array([1, 2, 3])
array_b = np.array([4, 5, 6])

soma = array_a + array_b
multiplicacao = array_a * 3

print(f"Soma: {soma}")
print(f"Multiplicação: {multiplicacao}")
```

```python
anos = [2020, 2021, 2022, 2023, 2024]
vendas = [130, 110, 120, 150, 120]

plt.plot(anos, vendas)
plt.title('Vendas ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Vendas (em milhões)')
plt.grid(True)
plt.show()
```

```python
produtos = ['Produto A', 'Produto B', 'Produto C', 'Produto D']
quantidades = [50, 80, 65, 60]

plt.bar(produtos, quantidades)
plt.title('Quantidade de Produtos (Estoque)')
plt.xlabel('Produtos')
plt.ylabel('Quantidades')
plt.show()
```

```python
np.random.seed(42)
x = np.random.rand(50) * 10
y = 2 * x + np.random.randn(50) * 2

plt.scatter(x, y)
plt.title('Relação entre X e Y')
plt.xlabel('Variável X')
plt.ylabel('Variável Y')
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y_seno = np.sin(x)
y_cossseno = np.cos(x)

plt.plot(x, y_seno, label='Seno')
plt.plot(x, y_cossseno, label='Cosseno')
plt.title('Funções Seno e Cosseno')
plt.xlabel('Ângulo [rad]')
plt.ylabel('Valor')
plt.legend()
plt.grid(True)
plt.show()
```

```python
dados = np.random.normal(loc=0, scale=1, size=1000)

plt.hist(dados, bins=30, color='skyblue', edgecolor='black')

plt.title('Histograma de uma Distribuição Normal')
plt.xlabel('Valores')
plt.ylabel('Frequência')
plt.show()
```

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

fig.suptitle('Múltiplos Gráficos (Subplots)', fontsize=16)

axs[0, 0].plot(x, np.sin(x), color='blue')
axs[0, 0].set_title('Seno')
axs[0, 0].grid(True)

axs[0, 1].plot(x, np.cos(x), color='orange')
axs[0, 1].set_title('Cosseno')
axs[0, 1].grid(True)

axs[1, 0].plot(x, np.tan(x), color='green')
axs[1, 0].set_title('Tangente')
axs[1, 0].set_ylim(-5, 5)
axs[1, 0].grid(True)

axs[1, 1].plot(x, x**2, color='purple')
axs[1, 1].set_title('Parábola (x^2)')
axs[1, 1].grid(True)

plt.tight_layout(rect=[0, 0, 1, 0.96])
axs[0, 1].show()
```

```python
categorias = ['Tecnologia', 'Saúde', 'Educação', 'Transporte']
valores = [45, 25, 15, 15]

explode = (0.1, 0, 0, 0)

plt.pie(valores, explode=explode, labels=categorias, autopct='%1.1f%%', shadow=True, startangle=90)

plt.axis('equal')

plt.title('Distribuição de Investimentos por Setor')
plt.show()
```
