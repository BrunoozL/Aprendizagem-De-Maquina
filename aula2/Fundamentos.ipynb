## Fundamentos de Python para Aprendizagem de Máquina

```python
print("Hello World!")
print("Olá Mundo!")
```

```python
# Lista (Mutável)
frutas = ["maçã", "banana", "cereja"]
frutas.append("uva")

# Tupla (Imutável)
coordenadas = (10, "Vini")

# Dicionário (Chave-Valor)
aluno = {"nome": "Bruno", "curso": "Tecnologia", "nota": 9.7}

# Interpolação ('f' no começo das strings)
print(f"Lista: {frutas}")
print(f"Dicionário: {aluno['nome']} tirou {aluno['nota']}")
```

```python
# Condicional
media = 7.5

if media >= 6:
    resultado = "Aprovado"
else:
    resultado = "Exame"

# Loop For
numeros = [1, 2, 3, 4, 5]
quadrados = [x**2 for x in numeros]

print(f"Resultado: {resultado}")
print(f"Quadrados: {quadrados}")

print("\nLista de Números (jeito longo)")
quadrados = []
for x in numeros:
    quadrados.append(x**2)
print(quadrados)
```

```python
def calcular_bonus(salario, percentual=10):
    return salario + (salario * percentual / 100)

# Função Lambda (Anônima)
dobro = lambda x: x * 2

print(f"Bônus: {calcular_bonus(3000)}")
print(f"Dobro de 5: {dobro(5)}")
```
