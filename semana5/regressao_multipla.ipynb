```python
import numpy as np

dados = np.genfromtxt('kc_house_data.csv', delimiter=',', skip_header=1, usecols=(2, 3, 4, 14), dtype=str)
dados = np.char.strip(dados, '"').astype(float) # Força os dados a virem do tipo float para poder fazer o cálculo

y = dados[:, 0]
X_features = dados[:, 1:]

m = X_features.shape[0]
X = np.c_[np.ones(m), X_features]

X_T = X.T

beta = np.linalg.inv(X_T @ X) @ X_T @ y

print("=== Coeficientes Descobertos ===")
print(f"Intercepto (Base 0): {beta[0]:.2f}")
print(f"Peso para Quartos (Base 1): {beta[1]:.2f}")
print(f"Peso para Banheiros (Base 2): {beta[2]:.2f}")
#print(f"Peso para Área/sqft (Base 3): {beta[3]:.2f}")
print(f"Peso para Área/sqft (Base 3): {beta[3]:.2f}\n")

casa_teste = np.array([1, 3, 1, 1955])
preco_previsto = casa_teste @ beta

print("=== Teste de Previsão ===")
print(f"Preço REAL da casa 1: $ {y[0]:.2f}")
print(f"Preço PREVISTO pelo nosso modelo: $ {preco_previsto:.2f}")
```
